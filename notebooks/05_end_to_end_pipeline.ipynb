## 2. Test Live FastAPI Endpoint (Indian Hospital Patient)


```python
sample_indian_patient = {
    "age": 52.0,
    "gender": 1,
    "chestpain": 0,
    "restingBP": 150.0,
    "serumcholestrol": 240.0,
    "fastingbloodsugar": 0,
    "restingrelectro": 1,
    "maxheartrate": 155.0,
    "exerciseangia": 1,
    "oldpeak": 1.8,
    "slope": 2,
    "noofmajorvessels": 1
}

try:
    with httpx.Client(timeout=10.0) as client:
        resp = client.post("http://127.0.0.1:8000/predict/india", json=sample_indian_patient)
        print("Status Code:", resp.status_code)
        print("Response JSON:")
        print(resp.json())
except Exception as e:
    print("FastAPI server offline or error:", e)
```
